## Hypothesis Testing Results

---

### **General Relationship Between PM2.5 and DALY**

Before testing specific hypotheses, I checked whether PM2.5 and DALY are related in general.  
Across the whole dataset, the correlation between PM2.5 and DALY per 100k is very strong and positive.  
This means that higher pollution levels are consistently linked to higher health burden.

This overall pattern gives a good starting point for the more detailed subgroup analyses below.

---

### **Hypothesis 1 — Age changes how strongly PM2.5 affects DALY**

**Idea:**  
Older people might be more affected by PM2.5 than younger people.

**Result:**  
This hypothesis is supported.  
Children (<15) show the weakest increase in DALY when PM2.5 rises.  
Adults have a stronger increase, and older adults (50+) have the steepest rise.  

In other words, **PM2.5 affects everyone, but especially older age groups**.  
The figure below shows these different slopes for each age category.

---

### **Hypothesis 2 — Different diseases respond differently to PM2.5**

**Idea:**  
Not all diseases should react to air pollution in the same way.

**Result:**  
This hypothesis is also supported.  
Lung cancer has the strongest PM2.5 sensitivity (the highest slope).  
Asthma and COPD show smaller, but still positive, increases in DALY as pollution rises.  

So **some diseases are much more strongly linked to PM2.5 than others**, and this helps explain the overall health burden.

---

The two figures below illustrate these main findings.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = get_clean_data("DataExtract.csv")
df = df[
    (df["pollutant"] == "PM2.5") &
    (df["category"] == "Total burden of disease") &
    (df["health_indicator"].str.contains("DALY", case=False, na=False)) &
    (df["pollution_avg"] > 0) &
    (df["value_per_100k"] > 0)
].copy()

df = df[df["pollution_avg"] <= df["pollution_avg"].quantile(0.99)]
df = df[df["value_per_100k"] <= df["value_per_100k"].quantile(0.99)]

df["age_num"] = df["age_group"].str.extract(r"(\d+)").astype(float).fillna(0)

df["age_class"] = pd.cut(
    df["age_num"],
    bins=[-1, 15, 50, 120],
    labels=["Children (<15)", "Adults (15–50)", "Older adults (50+)"]
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="pollution_avg",
    y="value_per_100k",
    hue="age_class",
    alpha=0.35
)


for group, sub in df.groupby("age_class", observed=False):
    if len(sub) > 2:      
        sns.regplot(
            data=sub,
            x="pollution_avg",
            y="value_per_100k",
            scatter=False,
            label=f"{group} trend"
        )

plt.title("Age × PM2.5 interaction")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("DALY per 100k population")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

top3 = df["outcome"].value_counts().index[:3]

slopes = []
for oc in top3:
    sub = df[df["outcome"] == oc]
    slope, b = np.polyfit(sub["pollution_avg"], sub["value_per_100k"], 1)
    slopes.append(slope)

slopes_df = pd.DataFrame({
    "Outcome": top3,
    "Slope": slopes
})

plt.figure(figsize=(8, 5))

sns.barplot(
    data=slopes_df,
    x="Outcome",
    y="Slope",
    hue="Outcome",        
    palette="Set2",
    legend=False         
)

for i, v in enumerate(slopes_df["Slope"]):
    plt.text(i, v + 0.4, f"{v:.1f}", ha="center", fontsize=10)

plt.title("Disease-specific PM2.5 sensitivity")
plt.ylabel("Slope (Δ DALY/100k per +1 µg/m³ PM2.5)")
plt.xlabel("")
plt.tight_layout()
plt.show()
```
